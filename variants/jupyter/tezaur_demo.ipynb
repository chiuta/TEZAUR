# TEZAUR în Jupyter

Pornește serverul (`tezaur serve`) și rulează celulele. Biblioteca și modelul rămân pe calculatorul tău.

```python
from tezaur_client import Tezaur
t = Tezaur('http://127.0.0.1:8787', token=None, workspace='default')
t.model, t.api
```

```python
# Adaugă un fișier (PDF-urile scanate sunt citite cu OCR pe server)
# t.add('/cale/catre/carte.pdf')
for d in t.sources():
    print(d['title'], '—', d['chunks'], 'pasaje')
```

```python
hits = t.search('sarcina probei', k=5)
for h in hits:
    print(h['n'], h['label'], f"({h['cov']:.0%} din termeni)")
    print('   ', h['text'][:160].replace('\n', ' '))
```

```python
r = t.ask('Cine poartă sarcina probei?', lang='română')
print(r['answer'])
print(r['warnings'])
```
